```python
import numpy as np
```

```python
student_marks = [85, 90, 78, 92, 88]

print("Student Marks:", student_marks)
```

```python
def calculate_mean(values):
    return sum(values) / len(values)

mean = calculate_mean(student_marks)
print("Mean:", mean)
```

```python
def find_maximum(values):
    return max(values)

print("Maximum:", find_maximum(student_marks))
```

```python
def find_minimum(values):
    return min(values)

print("Minimum:", find_minimum(student_marks))
```

```python
def calculate_percentage(obtained, total):
    return (obtained / total) * 100

percentage = calculate_percentage(430, 500)
print("Percentage:", percentage, "%")
```

```python
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

print("Grade:", calculate_grade(85))
```

```python
student_marks = [85, 90, 78, 92, 88]

print("Student Marks:", student_marks)
print("Mean:", calculate_mean(student_marks))
print("Maximum:", find_maximum(student_marks))
print("Minimum:", find_minimum(student_marks))
print("Percentage:", calculate_percentage(430, 500))
print("Grade:", calculate_grade(85))
```
